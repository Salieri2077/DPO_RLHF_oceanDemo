```python
    from transformers import AutoModelForCausalLM, AutoTokenizer
    
    model_name = "/root/autodl-tmp/Qwen2.5-7B-Instruct/"
    model = AutoModelForCausalLM.from_pretrained(
        model_name,
        torch_dtype="auto",
        device_map="auto"
    )
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    
    prompt = "Give me a short introduction to large language model."
    messages = [
        {"role": "system", "content": "You are Qwen, created by Alibaba Cloud. You are a helpful assistant."},
        {"role": "user", "content": prompt}
    ]
    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )
    model_inputs = tokenizer([text], return_tensors="pt").to(model.device)
    
    generated_ids = model.generate(
        **model_inputs,
        max_new_tokens=512
    )
    generated_ids = [
        output_ids[len(input_ids):] for input_ids, output_ids in zip(model_inputs.input_ids, generated_ids)
    ]
    
    response = tokenizer.batch_decode(generated_ids, skip_special_tokens=True)[0]
```

**向LLaMA-Factory/data/dataset_info.json中插入内容**

```python
import json
from datasets import load_dataset
import os

# LlamaFactory 的 data 根目录
LF_DATA_DIR = "/root/autodl-tmp/target_1/LLaMA-Factory/data"
# 配置文件路径 (dataset_info.json) <--- 这里修正了！
INFO_FILE_PATH = os.path.join(LF_DATA_DIR, "dataset_info.json")

# 定义配置内容
new_dataset_config = {
    "file_name": "ocean_dpo_data.json",
    "ranking": True,
    "formatting": "sharegpt", 
    "columns": {
    "messages": "conversations",
    "chosen": "chosen",
    "rejected": "rejected"
  }
}

# 读取并更新 dataset_info.json
current_config = {}
if os.path.exists(INFO_FILE_PATH):
    try:
        with open(INFO_FILE_PATH, 'r', encoding='utf-8') as f:
            current_config = json.load(f)
    except json.JSONDecodeError:
        print("⚠️ 原配置文件损坏，将新建。")
else:
    print("⚠️ 未找到 dataset_info.json，将新建。")

# 插入配置
# current_config["ocean_data"] = new_dataset_config
current_config["ocean_dpo"] = new_dataset_config

# 保存
with open(INFO_FILE_PATH, 'w', encoding='utf-8') as f:
    json.dump(current_config, f, ensure_ascii=False, indent=2)

print(f"✅ 注册成功！配置已写入: {INFO_FILE_PATH}")
print("🎉 一切就绪！现在可以运行 llamafactory-cli 的DPO train 命令了！")
```
